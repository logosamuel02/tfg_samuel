```python
from export import Config
import preprocess as pre
import visualization.plots as plo

def conversion_svg_figure(fig):
    html_fig: str = fig.to_html(full_html=True)
    html_fig = html_fig.replace("PNG", "SVG", 1)
    html_fig = html_fig.replace("png", "svg", 3)
    return html_fig

def convert_to_html(fig):
    html = conversion_svg_figure(fig)
    with open("my_plot.html", "w", encoding="utf-8") as file:
        file.write(html)

fig = plo.create_bubble_plot()
```
